```python
def is_even(number):
    if(number%2 ==0):
        print("Even Number")
    else:
        print("Odd Number")
    return

is_even(int(input("Enter the number: ")))
```

```python
def calculate_total(discount_percentage,acutal_price):
    return discount_percentage*acutal_price/100

Total_price=calculate_total(int(input("Enter The Discount Percentage: ")),int(input("Enter The Acutal Price")))

print(f"Total Amount To pay: {Total_price}")

```

```python
def count_vowels(Text):
    vowels=["A","a","E","e","I","i","O","o","U","u"]
    count=0
    for char in Text:
        if (char in vowels):
            count+=1
    return count
    
word=input("Enter the word: ")
total=count_vowels(word)
print(f"Total Vowels: {total}")
```
