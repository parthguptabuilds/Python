.startswith(prefix):- it return the boolean value

```python
name="Parth"
s=name.startswith("P")
print(s)
```

.endswith(suffix):-it return the boolean value

```python
name="helloparth"
s=name.endswith("th")
print(s)
```

.isalpha():-it return the boolean value

```python
name="Parth"
aplha=name.isalpha()
print(aplha)
```

.isdigit():-it return the boolean value

```python
number="1111111"
digi=number.isdigit()
print(digi)
```

.isalnum():-it return the boolean value

```python
username="parth1234"
s=username.isalnum()
print(s)
```

.lower()

```python
name="PARTH"
s=name.lower()
print(s)
```

.upper()

```python
name="parth"
s=name.upper()
print(s)
```

.title()

```python
name="hello world"
s=name.title()
print(s)
```

```python
name="_"
s=name.join(["red","ffff","dsdsdd"])
print(s)
```
